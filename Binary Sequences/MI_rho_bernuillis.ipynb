```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Function to calculate mutual information
def mutual_information(rho, p=0.5, q=0.5):
    p00 = (1 - p) * (1 - q) + rho * np.sqrt(p * (1 - p) * q * (1 - q))
    p01 = q * (1 - p) - rho * np.sqrt(p * (1 - p) * q * (1 - q))
    p10 = p * (1 - q) - rho * np.sqrt(p * (1 - p) * q * (1 - q))
    p11 = rho * np.sqrt(p * (1 - p) * q * (1 - q)) + p * q

    # Calculate individual probabilities
    pA0 = 1 - p
    pA1 = p
    pB0 = 1 - q
    pB1 = q

    # Ensure the probabilities are non-negative and valid
    assert np.all(p00 >= 0) and np.all(p01 >= 0) and np.all(p10 >= 0) and np.all(p11 >= 0), "Invalid probabilities"

    # Calculate mutual information
    mi = (
        p00 * np.log2(p00 / (pA0 * pB0)) +
        p01 * np.log2(p01 / (pA0 * pB1)) +
        p10 * np.log2(p10 / (pA1 * pB0)) +
        p11 * np.log2(p11 / (pA1 * pB1))
    )
    
    return mi

# Generate rho values from -1 to 1
rho_values = np.linspace(-1, 1, 500)

# Calculate mutual information for each rho
mi_values = [mutual_information(rho) for rho in rho_values]

# Plot the mutual information as a function of rho
plt.figure(figsize=(10, 6))
plt.plot(rho_values, mi_values, label='Mutual Information')
plt.xlabel(r'$\rho$')
plt.ylabel('Mutual Information $I(A_i; B_i)$')
plt.title('Mutual Information vs. Correlation Coefficient')
plt.legend()
plt.grid(True)
plt.show()
```
